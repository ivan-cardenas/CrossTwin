```python
from owslib.wfs import WebFeatureService
```

```python
getCapabilities = 'service=WFS&request=GetCapabilities'
url = 'https://service.pdok.nl/lv/bag/wfs/v2_0?'+getCapabilities
```

```python
wfs = WebFeatureService(url)
```

```python
for name, layer in wfs.contents.items():
    print(f"Name: {name}, Layer: {layer}")
    print(f"Title: {layer.title}")
    print("-"*50)
```

```python
response = wfs.get_schema('bag:pand')
import pprint 

pprint.pprint(response["properties"])

```

```python
# Force WFS 2.0.0 explicitly — owslib is defaulting to 1.0.0, which has this propertyname bug
wfs = WebFeatureService(url, version='2.0.0')

response2 = wfs.getfeature(typename='bag:pand', maxfeatures=1, startindex=35)

import geopandas as gpd
gdf = gpd.read_file(response2)
# print(gdf.dtypes)
gdf.iloc[0]
```
